```python
from openai import OpenAI
import os

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

# Toy data — no external dataset needed
reviews = [
    "The banana pudding was really tasty!",
    "Terrible service, never coming back.",
    "It was okay, nothing special."
]

def llm_sentiment(review):
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[{
            "role": "user",
            "content": f"Classify the following review as Positive, Negative, or Neutral. Reply with one word only.\n\nReview: {review}"
        }],
        temperature=0
    )
    return response.choices[0].message.content.strip()

for review in reviews:
    print(f"Review: {review}")
    print(f"Sentiment: {llm_sentiment(review)}\n")
```
